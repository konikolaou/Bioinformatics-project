```python
# Συνάρτηση που ψάχνει για κωδική αλληλουχία (Open Reading Frame - ORF) σε
# ένα τμήμα DNA από προκαρυωτικό κύτταρο.
# Ως είσοδος από τον χρήστη δίνεται μια πολυνουκλεοτιδική αλυσίδα.
# Σε αυτή υπάρχει δυνατότητα να καθοριστεί από τον χρήστη ο προσανατολισμός
# 5'->3' ή 3'->5', αν είναι γνωστός.

# Ο γενετικός κώδικας είναι κώδικας τριπλέτας, συνεχής και μη-επικαλυπτόμενος.
# Έχει κωδικόνιο έναρξης και κωδικόνιο λήξης.
# Στο προκαρυωτικό κύτταρο δεν υπάρχουν ενδιάμεσες αλληλουχίες DNA σε μια κωδική
# αλληλουχία, οι οποίες δεν περιέχουν κωδικόνια και δεν κατά συνέπεια δε μεταφράζονται.
# Η ανάγνωση της γενετικής πληροφορίας όταν ξεκινήσει από το ATG (κωδικόνιο έναρξης)
# συνεχίζεται ανά τρία νουκλεοτίδια με συνεχή και μη-επικαλυπτόμενο τρόπο
# ως το κωδικόνιο λήξης (TAA, TGA, TAG), όπου και σταματάει.

#Δημιοργία κώδικα εύρεσης αναγνωστικού πλαισίου με τη συνάρτηση def()
def find_orf(sequence, label):
    start_codon = "ATG"
    stop_codons = ["TAA", "TGA", "TAG"]
    found_any = False

    print(f"\n--- Έλεγχος αλυσίδας: {label} ---")


#Μια προκαρυωτική αλληλουχία DNA, που περιέχει μια κωδική περιοχή ή
#όπως αλλιώς αναφέρεται στη βιβλιογραφία ένα ανοιχτό πλαίσιο ανάγνωσης
#(Open Reading Frame - ORF), έχει την ακόλουθη δομή: 5'αμετάφραστη περιοχή
#(χωρίς κωδικόνια), κωδική αλληλουχία (με τα κωδικόνια),
# 3'αμετάφραστη περιοχή χωρίς κωδικόνια).
# Για να προσδιορίσουμε την ύπαρξη ή όχι μιας κωδικής αλληλουχίας σε μια
# προκαρυωτική πολυνουκλεοτιδική αλυσίδα, ξεκινάμε την ανάγνωση όπως περιγράφηκε,
# α)από το πρώτο, β) από το δεύτερο και γ) από το τρίτο νουκλεοτίδιο από
# την αρχή της αλυσίδας. Ελέγχουμε τρία πλαίσια ανάγνωσης.
# Υπάρχει περίπτωση να βρεθούν δύο ή περισσότερα αναγνωστικά πλαίσια σε
# μια αλληλουχία, που να ικανοποιούν την παραπάνω συνθήκη.
# Μπορούμε να επιλέξουμε το πιθανότερο λαμβάνοντας υπόψη την αλληλουχία Kozak γύρω από το ATG
# που καθορίζει την αποτελεσματικότητα της μετάφρασης.
# Στη συγκεκριμένη συνάρτηση αυτό δεν έχει ληφθεί υπόψη.
# Δίνεται ως απάντηση το αναγνωστικό πλαίσιο που προκύπτει από το πρώτο ATG που εντοπίζεται


# Οδηγίες για την ανάγνωση στην πολυνουκλεοτιδική αλυσίδα των 3 πλαισίων ανάγνωσης (Reading Frames)
    for frame in range(3):
        print(f"Πλαίσιο ανάγνωσης {frame + 1}:")
        found_in_frame = False

        # Διαβάζουμε ανά 3 ξεκινώντας από το 'frame'
        for i in range(frame, len(sequence) - 2, 3):
            codon = sequence[i:i+3]

            if codon == start_codon:
                # Μόλις βρούμε ATG, ψάχνουμε για ένα από τα Stop Codon παρακάτω, με βήμα τριπλέτας
                for j in range(i, len(sequence) - 2, 3):
                    current_codon = sequence[j:j+3]
                    if current_codon in stop_codons:
                        coding_seq = sequence[i:j+3]
                        print(f"   [!] Βρέθηκε κωδικοποιούσα αλληλουχία: {coding_seq}")
                        # Εμφάνιση των κωδικονίων της κωδικής αλληλουχίας
                        codons = [coding_seq[k:k+3] for k in range(0, len(coding_seq), 3)]
                        print("      Κωδικόνια:", " - ".join(codons))
                        found_in_frame = True
                        found_any = True
                        break # Σταματάμε στο πρώτο Stop

                if found_in_frame: break # Όταν βρεθεί το πρώτο ATG του πλαισίου σταματάει τον έλεγχο για επόμενο

        if not found_in_frame:
            print("   [-] Δε βρέθηκε κωδικοποιούσα περιοχή σε αυτό το πλαίσιο.")

    return found_any

# Στις DNA αλληλουχίες η γενετική πληροφορία αποθηκεύεται 5'->3'
# Αν είναι γνωστός ο προσανατολισμός της αλυσίδας η ανάγνωση πρέπει να γίνει μόνο 5'->3'
# Εξετάζονται 3 διαφορετικά πλαίσια ανάγνωσης για κάθε αλυσίδα
# Αν δεν είναι γνωστός ο προσανατολισμός της αλυσίδας η ανάγνωση θα γίνει ξεκινώντας από κάθε ακρο ξεχωριστά
# Εξετάζονται 6 διαφορετικά πλαίσια ανάγνωσης για κάθε αλυσίδα

# Ο χρήστης θα ερωτηθεί αρχικά για την αλληλουχία του DNA και
# ακολούθως για τον προσανατολισμό της αλληλουχίας

# Εισαγωγή από τον χρήστη, αλληλουχίας DNA και ακολούθως προσανατολισμού αν είναι γνωστός
dna_input = input("Εισάγετε την αλληλουχία DNA: ").upper().strip()
orientation = input("Εισάγετε προσανατολισμό (5-3 ή 3-5) της δοθείσας αλυσίδας ή πατήστε Enter για άγνωστο προσανατολισμό: ").strip()

# Στο δίκλωνο προκαρυωτικό DNA η κωδικοποιούσα περιοχή μπορεί να βρίσκεται στην αλυσίδα που
# είναι συμπληρωματική αυτής που εισάγει ο χρήστης
# Επίσης, είναι δυνατόν να υπάρχει κωδικοποιούσα περιοχή και στις δύο αλυσίδες του δίκλωνου τμήματος DNA
# Επομένως πρέπει να προσδιοριστεί και να ελεγθεί και η συμπληρωματική αλυσίδα από αυτή που δίνει ο χρήστης
# Επίσης, αν υπάρχει η πληροφορία για τον προσανατολισμό της δοθείσας αλυσίδας η συμπληρωματική πρέπει να οριστεί ως αντιπαράλληλη

# Δημιουργία της αλληλουχίας για τη δεύτερη αλυσίδα, που είναι συμπληρωματική και αντιπαράλληλη της δοθείσας
mapping = str.maketrans("ATCG", "TAGC")
complementary = dna_input.translate(mapping)

# Ανάγνωση των αναγνωστικών πλαισίων όταν δίνεται ο προσανατολισμός της δοθείσας αλυσίδας
if orientation == "5-3":
    # Η δοθείσα έχει 5-3, οπότε η συμπληρωματική και αντιπαράλληλη έχει 3-5
    find_orf(dna_input, "Αρχική (5'->3')")
    find_orf(complementary[::-1], "Συμπληρωματική Αντιπαράλληλη (5'->3')")

elif orientation == "3-5":
    # Η δοθείσα έχει 3-5, οπότε η συμπληρωματική και αντιπαράλληλη έχει  5-3
    find_orf(dna_input[::-1], "Αρχική (5'->3')")
    find_orf(complementary, "Συμπληρωματική Αντιπαράλληλη (5'->3')")

else:
    # Αν δεν υπάρχει προσανατολισμός, ελέγχουμε τα πιθανά αναγνωστικά παλάισια
    # από κάθε άκρο (4 περιπτώσεις-3 αναγνωστικά πλαίσια για κάθε περίπτωση)
    print("\n[Προσοχή] Δεν δόθηκε προσανατολισμός. Έλεγχος από κάθε άκρο:")
    find_orf(dna_input, "Αρχική (από αριστερά προς δεξιά)")
    find_orf(dna_input[::-1], "Αρχική (από δεξιά προς αριστερά)")
    find_orf(complementary, "Συμπληρωματική (από αριστερά προς δεξιά)")
    find_orf(complementary[::-1], "Συμπληρωματική (από δεξιά προς αριστερά)")

```
